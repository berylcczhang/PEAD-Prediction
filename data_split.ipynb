```python
import pandas as pd
```

```python
data = pd.read_csv('data/full_dataset.csv')
data.head()
```

```python
data.shape
```

```python
n = len(data)
train_end = int(n * 0.70)
val_end = int(n * 0.85)

# Split chronologically
train_df = data.iloc[:train_end].copy()
val_df = data.iloc[train_end:val_end].copy()
test_df = data.iloc[val_end:].copy()
```

```python
# Save the split datasets to CSV files
train_df.to_csv('data/training_events.csv', index=False)
val_df.to_csv('data/validation_events.csv', index=False)
test_df.to_csv('data/testing_events.csv', index=False)
```

```python
print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

print("\nDate ranges:")
print("Train:", train_df["date"].min(), "→", train_df["date"].max())
print("Validation:", val_df["date"].min(), "→", val_df["date"].max())
print("Test:", test_df["date"].min(), "→", test_df["date"].max())
```
