```python
import pandas as pd
```

```python
df_trscrpt = pd.read_csv('earnings_transcripts_sorted.csv')
df_trscrpt.head()
```

```python
df_mrkt = pd.read_csv('marketdata_to_events.csv')
df_mrkt.head()
```

```python
df_trscrpt.shape, df_mrkt.shape
```

```python
#split train, validation, and test sets
n_events = len(df_trscrpt)
train_end = int(n_events * 0.70)
val_end = int(n_events * 0.85)

train_events = df_trscrpt.iloc[:train_end].copy()
val_events = df_trscrpt.iloc[train_end:val_end].copy()
test_events = df_trscrpt.iloc[val_end:].copy()

train_market = df_mrkt[df_mrkt["event_id"].isin(train_events["event_id"])].copy()
val_market = df_mrkt[df_mrkt["event_id"].isin(val_events["event_id"])].copy()
test_market = df_mrkt[df_mrkt["event_id"].isin(test_events["event_id"])].copy()

print("Events:")
print(f"Train:      {len(train_events):,}")
print(f"Validation: {len(val_events):,}")
print(f"Test:       {len(test_events):,}")

print("\nDate ranges:")
print(
    f"Train:      {train_events['date'].min()} → "
    f"{train_events['date'].max()}"
)
print(
    f"Validation: {val_events['date'].min()} → "
    f"{val_events['date'].max()}"
)
print(
    f"Test:       {test_events['date'].min()} → "
    f"{test_events['date'].max()}"
)

print("\nMarket rows:")
print(f"Train:      {len(train_market):,}")
print(f"Validation: {len(val_market):,}")
print(f"Test:       {len(test_market):,}")
```

```python
# save the split datasets to CSV files
train_events.to_csv('data/train_events.csv', index=False)
val_events.to_csv('data/val_events.csv', index=False)
test_events.to_csv('data/test_events.csv', index=False)
train_market.to_csv('data/train_market.csv', index=False)
val_market.to_csv('data/val_market.csv', index=False)
test_market.to_csv('data/test_market.csv', index=False)
```
